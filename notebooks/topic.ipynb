```python
import pandas as pd
from konlpy.tag import Okt
from gensim.corpora.dictionary import Dictionary
from gensim.models import LdaModel
from sklearn.feature_extraction.text import TfidfVectorizer
import re
```

```python
df = pd.read_csv('workspace/clustering.csv')
df
```

```python
# df['word'].apply(lambda x: [word for word in eval(x) if word not in stopwords and len(word) > 1])

stopwords = ['먹는', '있어요', '부터', '있어', '특히', '닭도리탕', '소정', '새우', '복어', '치킨', '와인', '낙성대', '낙성대역', 'KRW', '스테이크', '식사', '니당', '서울', '버거', '에도', '두부김치', '족발', '먹고', '다음', '국물']

df['word'] = df['word'].apply(lambda x: [word for word in x if word not in stopwords and len(word) > 1])
```

```python
for c in df['c'].unique():
    print(c)
    dictionary = Dictionary(df[df['c'] == c]['word'])
    corpus = [dictionary.doc2bow(text) for text in df[df['c'] == c]['word']]
    lda = LdaModel(corpus=corpus, num_topics=1, id2word=dictionary, passes=15)
    topics = lda.print_topics(num_words=5)
    print(topics)
```
