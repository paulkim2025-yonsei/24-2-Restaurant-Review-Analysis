```python
import numpy as np

f = np.load('workspace/ft.npy')
label = np.load('workspace/label.npy')
```

```python
import warnings
warnings.filterwarnings('ignore')
```

```python
f.shape
```

```python
from sklearn.preprocessing import MinMaxScaler

scaler = MinMaxScaler()
data = f
data = scaler.fit_transform(data)
```

```python
from sklearn.decomposition import PCA

pca = PCA(random_state=0)
X_p = pca.fit_transform(data)
```

```python
each_variance_ratio = pca.explained_variance_ratio_
cumsum = np.cumsum(each_variance_ratio)
cumsum
```

```python
np.argmax(cumsum >= 0.8) + 1
```

```python
pca = PCA(n_components=3,random_state=0)
pca_out = pca.fit(data)
X_p = pca.fit_transform(data)
```

```python
X_p.shape
```

```python
import matplotlib.pyplot as plt

plt.scatter(x=X_p[:, 0],y=X_p[:, 1], c=label)
```

```python
fig = plt.figure(figsize=(10,8))
ax = fig.add_subplot(projection = '3d')
ax.scatter(X_p[:, 0], X_p[:, 1], X_p[:, 2], c = label)
```
